### List vs numpy array

```python
py_list = [1,2,3]

print("Python list multiplication", py_list *2 )


np_array= np.array([1,3,6]) # element wise multiplication
print("np array  multiplication", np_array *2 )
```

```python
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List operation time: ", time.time()-start)

start = time.time()
np_array = np.array(1000000) * 2
print("\n numpy array operation time: ", time.time()-start)
```

### ARRAY RESHAPING

```python
arr = np.arange(12)
print(arr)
```

```python
print("shape ", arr.shape)
print("Dimension", arr.ndim)

print("Size ", arr.size)

print("shape ", arr.dtype)
```

```python
reshaped = arr.reshape((2,6))
print("reshaped array", reshaped)
```

```python
flattend = reshaped.flatten()
print("flattend array", flattend)
print("reshaped array", reshaped)


```

```python
###### revel return view intead of copy
releved = reshaped.ravel()
print("releved array", releved)
```

### transpose matrixs

```python
transpose = reshaped.T
print("transpose ", transpose)
```
