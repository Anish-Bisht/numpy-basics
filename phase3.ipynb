##### Advance operation with Bussiness examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python


# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("==== Zomato sales analysis")

print("\n shape of the data", sales_data.shape)
print("first 3 restruant data", sales_data[0:3])
print("first 3 restruant data", sales_data[:, 1: ]) # removing ids
```

```python
#total_sales_per_year 
print(np.sum(sales_data, axis=0)) 
n =np.sum(sales_data, axis=0)
n = n[1:]
print(n)
#year total
print("ssfsdf",np.sum(sales_data[:,1:], axis=1)) 

```

```python
#minmum sales 
min_sales= np.min(sales_data[:,1:],axis=0)
print("minimum sales ", min_sales)
```

```python
# Maximum sales per year 
max_sales= np.max(sales_data[:,1:],axis=0)
print("max sales per year sales ", max_sales)
```

```python
## Averge sales per restrauant
avg_sales = np.average(sales_data[:,1:],axis=1)
## another way avg_sales = np.mean(sales_data[:,1:],axis=1)
print(avg_sales)
```

```python
#Cumleative sales
cumcum = np.cumsum(sales_data[:,1:], axis=1)
#print(cumcum)

plt.Figure(figsize=(8,5))
plt.plot(np.mean(cumcum,axis=0))
plt.title = ("Average cumulative  sales accross all restruant")
plt.xlabel = ("Years")
plt.ylabel = ("Sales")
plt.grid(True)
plt.show()

```

```python
# vectors 
vector1 = np.array([1,2,3])
vector2 = np.array([6,5,3])
additon = vector1 + vector2
print("additon ", additon)

mul = vector1 * vector2
print("mul", mul)

#dot prodcut
dot_product = np.dot(vector1,vector2)
print("dot product",dot_product)

#angle between two vec
angle = np.arccos(np.dot(vector1,vector2)/ (np.linalg.norm(vector1)* np.linalg.norm(vector2)))
print("angle", angle) 


```

```python
#Performe a operation on all the items of vector or tensor or matrics called vectorized operation

resturant_types = np.array(['biryani', 'chinese', 'pizza', 'burger', 'cafe'])

vectorized_operation = np.vectorize(str.upper)
print("vectorized upper", vectorized_operation(resturant_types))
```
