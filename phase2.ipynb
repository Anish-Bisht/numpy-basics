## Numpy Array Operation

```python
import numpy as np

```

```python
arr = np.array(range(10))
print("arr", arr)
print("Slicing", arr[2:7])
print("with step slicing", arr[2:10: 3])
print("negative indexing ", arr[-2])
print("negative step indexing ", arr[8:3:-3])


```

```python
arr_2d = np.array([[4,5,6,3], [23,65,34,5],[5,6,7,3]])
print("2d array",arr_2d) 

print("specific element ", arr_2d[1,2])
print("specific element ", arr_2d[1][2])
print("entire row elements ", arr_2d[1])
print("entire column elements ", arr_2d[:,1])


```

## Sorting

```python
unsorted_array = np.array([2,1,5,3,])
```

```python
sort_arr = np.sort(unsorted_array)
print("sorted array", sort_arr)

```

```python
arr_2d_unstored = np.array([[3,2,5], [4,2,3], [7,4,2],])
print("2d unstored array", arr_2d_unstored)

print("25sort array by column", np.sort(arr_2d_unstored, axis=0)) ## 0 is column wise sorting
print("25sort array by row", np.sort(arr_2d_unstored, axis=1)) ## 1 is row wise sorting



```

## Filtring

```python
numbers = np.array([1,2,3,4,5,6,24,1,67])
```

```python
even_number = numbers[numbers %2==0]
print("even number ", even_number)
```

##### Filter With Mask

```python
mask = numbers > 5
print("Number greater then 5", numbers[mask])
```

##### Fancy indexing vs np.where()

```python
indices = [0,2,4]
print("number indices ", numbers[indices])
```

```python
where_result = np.where(numbers >5)
print("np where cluase", numbers[where_result])
```

```python
conditon_array = np.where(numbers >5, numbers *2, numbers)
print(conditon_array)
```

Wherer with multiple condition, Above experssion works like this

if(numbers>5){
    numbers *2
}else{
    numbers
}

#### Adding And Removing Data

```python
arr1 = np.array([1,2,3])
arr2 = np.array([3,4,5])

add_arr = arr1 + arr2
print("add result", add_arr)
combined = np.concatenate([arr1,arr2])
print("combined result", combined)
```

##### Array Compatibility

```python
a= np.array([1,2,3])
b= np.array([4,5,6])
c= np.array([4,5,6])

print("compatibility shapes", a.shape == b.shape)
```

```python
orginal = np.array([[1,2],[3,4]])
new_row = np.array([[5,6]])

with_new_row = np.vstack((orginal, new_row)) ### vstack adds the row
print("original", orginal)
print("new_row", new_row)
print("with_new_row", with_new_row)

new_col = np.array([[7],[8]])
with_new_col = np.hstack((orginal, new_col))
print("original", orginal)
print("new_col", new_col)
print("with_new_col", with_new_col)
```
